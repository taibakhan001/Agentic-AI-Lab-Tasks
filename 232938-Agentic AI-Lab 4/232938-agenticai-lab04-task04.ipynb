```python
import pandas as pd

# Path to the dataset
DATA_PATH = "/kaggle/input/datasets/binovi/wholesale-customers-data-set/Wholesale customers data.csv"

# Load the dataset
df = pd.read_csv(DATA_PATH)

# Display dataset shape and first 5 rows
print(df.shape)
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Create scatter plot
plt.figure(figsize=(6, 5))

plt.scatter(
    df["Grocery"],
    df["Detergents_Paper"],
    s=15,
    alpha=0.6
)

# Add labels
plt.xlabel("Grocery Spending")
plt.ylabel("Detergents & Paper Spending")
plt.title("Grocery vs. Detergents & Paper Spending")

# Display the plot
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

# Select features
feature_cols = [
    "Fresh",
    "Milk",
    "Grocery",
    "Frozen",
    "Detergents_Paper",
    "Delicassen"
]

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[feature_cols].values)

print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans

# Create the K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Fit the model
kmeans.fit(X_scaled)
```

```python
# Assign each customer to a cluster
df["Cluster"] = kmeans.predict(X_scaled)

# Display the number of customers in each cluster
print(df["Cluster"].value_counts())
```

```python
# Compare clusters with customer channels
comparison = pd.crosstab(
    df["Cluster"],
    df["Channel"]
)

print(comparison)
```
