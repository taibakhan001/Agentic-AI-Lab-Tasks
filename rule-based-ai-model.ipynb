```python
def spam_classifier(email_text):

    spam_keywords = ["free", "winner", "urgent", "click here", "prize"]

    email_text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_text:
            count += 1

    if count >= 2:
        return "Spam", count
    else:
        return "Legitimate", count


email = "Congratulations! You are a winner! Claim your free prize now!"

result, count = spam_classifier(email)

print("Email:", email)
print("Spam keyword count:", count)
print("Classification:", result)
```
