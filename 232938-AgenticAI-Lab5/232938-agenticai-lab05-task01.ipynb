```python

import os

for root, dirs, files in os.walk("/kaggle/input"):
    for file in files:
        print(os.path.join(root, file))
```

```python

prompt_file = "/kaggle/input/datasets/ratthachat/writing-prompts/writingPrompts/train.wp_source"

with open(prompt_file, "r", encoding="utf-8") as f:
    prompts = [line.strip() for line in f if line.strip()]

print("Total prompts available:", len(prompts))

for i, prompt in enumerate(prompts[:5], start=1):
    print(f"\nPrompt {i}:\n{prompt}")
```

```python

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

print("GPT-2 model loaded successfully!")
```

```python

selected_prompts = prompts[:5]

for i, prompt in enumerate(selected_prompts, start=1):
    result = generator(
        prompt,
        max_new_tokens=100,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(f"\n{'='*60}")
    print(f"PROMPT {i}:")
    print(prompt)
    print(f"\nGENERATED COMPLETION {i}:")
    print(result[0]["generated_text"][len(prompt):])
```
